```python
#list also a mutable ,,its ,means data can changed like apend ,remove like this 
#[]brackets belogs to square brackets,used for list
fruits = ["apple", "banana", "cherry",'orange']
print(fruits)
 #accessing the first item
```

```python
print(fruits[0])
```

```python
#append fruit from the lsit 
fruits.append("grape")
print('After appending:', fruits)
```

```python
#remove an item from the list
fruits.remove("banana")
print('After removing banana:', fruits)
```

```python
#alterative :remove by index using pop() method
fruits.pop(1)  #removing the second item (index 1)
print('After popping index 1:', fruits)
```
